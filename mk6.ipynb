```python
import numpy as np
import matplotlib.pyplot as plt

# --- Definicje funkcji ---
def relu(x):
    return np.maximum(0, x)

def relu_grad(x):
    grad = np.zeros_like(x)
    grad[x > 0] = 1
    return grad

def leaky_relu(x, alpha=0.1):
    return np.where(x > 0, x, alpha * x)

def leaky_relu_grad(x, alpha=0.1):
    grad = np.ones_like(x)
    grad[x <= 0] = alpha
    return grad

# --- Dane wejściowe ---
x = np.linspace(-10, 10, 500)
y_relu = relu(x)
y_leaky = leaky_relu(x)
dy_relu = relu_grad(x)
dy_leaky = leaky_relu_grad(x)

# --- Wykres funkcji ---
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.plot(x, y_relu, label='ReLU', color='blue')
plt.plot(x, y_leaky, label='Leaky ReLU', color='green')
plt.title("Porównanie funkcji aktywacji")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()

# --- Wykres gradientów ---
plt.subplot(1,2,2)
plt.plot(x, dy_relu, label="ReLU'", color='red', linestyle='--')
plt.plot(x, dy_leaky, label="Leaky ReLU'", color='orange', linestyle='--')
plt.title("Gradient funkcji aktywacji")
plt.xlabel("x")
plt.ylabel("f'(x)")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
